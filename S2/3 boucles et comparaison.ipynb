# Boucles et comparaisons

## Boucles `for`

En programmation, on est souvent amené à répéter plusieurs fois une instruction. Les boucles permettent de réaliser cette tâche de manière **compacte et efficace**.

````python
animaux = ["girafe", "tigre", "singe", "souris"]
for animal in animaux:
    print(animal)
````

## Corps de la boucle et indentation

- `animaux` est une **liste**, donc un objet **itérable**.
- `animal` est une **chaîne de caractères**, car chaque élément de `animaux` est une chaîne.

Une boucle `for` permet d’itérer sur **n’importe quel objet séquentiel** (liste, chaîne, tuple…).

> ⚠️ En Python, **l’indentation** (le décalage vers la droite) permet de définir **le bloc d’instructions** à répéter.

````python
for animal in animaux:
print(animal)
````

> La fonction `range()` permet d’itérer automatiquement sur une suite d’entiers. Elle est **optimisée pour les boucles**.

````python
for i in range(4):
    print(i)
````

````python
for i, animal in enumerate(animaux):
    print(f"l'animal numero {i} est un(e) {animal:>10s}.")
````

## Boucles `while`

Une alternative aux boucles `for` est la boucle `while`.  
Elle exécute un bloc d’instructions **tant qu’une condition est vraie**.

````python
i = 1
while i <= 4:
    print(i)
    i += 1
````

Une boucle `while` nécessite en général **trois éléments** pour fonctionner correctement :

1. Initialisation de la variable d’itération (avant la boucle).
2. Condition logique à tester (dans l’en-tête du `while`).
3. Mise à jour de la variable (dans le corps de la boucle).

⚠️ **Attention** : si vous oubliez d’incrémenter la variable, cela crée une **boucle infinie** !

````python
i = 0
while i < 10:
    print("Le Python c'est cool !")
````

### Boucle `while` avec `input()`

On peut combiner la boucle `while` avec la fonction `input()` pour demander à l’utilisateur une valeur jusqu’à ce qu’elle respecte une condition.

````python
i = 0
while i < 10:
    reponse = input("Entrez un entier supérieur à 10 : ")
    i = int(reponse)
````
