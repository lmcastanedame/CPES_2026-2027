# Tests

Les **tests** permettent à Python de prendre des décisions. On utilise l’instruction `if` avec une **comparaison**.

````python
x = 2
if x == 2:
    print("Le test est vrai !")
````

````python
x = "souris"
if x == "tigre":
    print("Le test est vrai !")
````

> - Le bloc d’instructions du `if` doit être **indenté**.
> - Le `if` se termine par deux-points `:`, comme pour les boucles `for` et `while`.

## Tests à plusieurs cas

Il est possible de tester si une condition est **vraie** ou **fausse** avec une seule instruction `if`. On peut ajouter un bloc `else` pour le cas contraire.

````python
x = 2
if x == 2:
    print("Le test est vrai !")
else:
    print("Le test est faux !")

x = 3
if x == 2:
    print("Le test est vrai !")
else:
    print("Le test est faux !")
````

> On peut aussi enchaîner plusieurs tests avec `if`, `elif` et `else`, ce qui permet de tester plusieurs valeurs.

````python
import random

base = random.choice(["a", "b", "c", "d"])

if base == "a":
    print("La base est a !")
elif base == "b":
    print("La base est b !")
elif base == "c":
    print("La base est c !")
elif base == "d":
    print("La base est d !")
````

> Python exécute **la première condition vraie** et ignore les suivantes.  S’il n’y en a aucune, on peut ajouter un bloc `else`.

## Importance de l'indentation

L'**indentation** est cruciale en Python. Elle détermine quelles instructions font partie d'un bloc conditionnel ou d'une boucle.

````python
nombres = [4, 5, 6]
for nb in nombres:
    if nb == 5:
        print("Le test est vrai")
        print(f"car la variable nb vaut {nb}")
````

````python
nombres = [4, 5, 6]
for nb in nombres:
    if nb == 5:
        print("Le test est vrai")
    print(f"car la variable nb vaut {nb}")
````

1. 
    - Les deux instructions sont **dans le bloc `if`** (elles sont indentées deux fois).  
    - Elles ne s'exécutent **que si `nb == 5`**.

2. 
    - Dans le **Code 2**, seule la première instruction est dans le bloc `if`.  
    - La seconde (`print(...)`) est indentée **une seule fois**, donc elle **s'exécute à chaque tour de boucle**.

## Tests multiples

Les **tests multiples** permettent de vérifier plusieurs conditions en même temps grâce aux **opérateurs booléens**.   

- `or` → (OU logique)
- `and` → (ET logique)

Voici un rappel :

**Table de vérité pour `or` :**

| Condition 1 | Opérateur | Condition 2 | Résultat |
|-------------|-----------|-------------|----------|
| Vrai        | OU        | Vrai        | Vrai     |
| Vrai        | OU        | Faux        | Vrai     |
| Faux        | OU        | Vrai        | Vrai     |
| Faux        | OU        | Faux        | Faux     |

**Table de vérité pour `and` :**

| Condition 1 | Opérateur | Condition 2 | Résultat |
|-------------|-----------|-------------|----------|
| Vrai        | ET        | Vrai        | Vrai     |
| Vrai        | ET        | Faux        | Faux     |
| Faux        | ET        | Vrai        | Faux     |
| Faux        | ET        | Faux        | Faux     |

> ⚠️ Ils doivent être écrits en **minuscules**.

````python
x = 2
y = 2

if x == 2 and y == 2:
    print("le test est vrai")
````

````python
if x == 2:
    if y == 2:
        print("le test est vrai")
````

````python
print(not True)
print(not False)
print(not (True and True))
````

## Instructions `break` et `continue`

Ces deux instructions modifient le comportement d’une boucle (`for` ou `while`) à l’aide d’un test.

> L'instruction `break` stoppe la boucle en cours :

````python
for nombre in range(4):
    if nombre > 1:
        break
    print(nombre)
````

> L'instruction `continue` saute à l’itération suivante, sans exécuter la suite du bloc d’instructions :

````python
for nombre in range(4):
    if nombre == 2:
        continue
    print(nombre)
````

## Tests de valeur sur des `floats`

Quand on souhaite tester la valeur d’une variable de type `float`, le premier réflexe serait d’utiliser l’opérateur d’égalité.

````python
1/10 == 0.1
````

````python
print((3 - 2.7) == 0.3)
print(3 - 2.7)
````

````python
print(0.3)
print(f"{0.3:.5f}")
print(f"{0.3:.60f}")
````

> Utilisez plutôt une **tolérance** avec `abs(val1 - val2) < epsilon`, par exemple `1e-10` ou `1e-15`.

## Comparer des `floats` avec une tolérance (`delta`)

Pour éviter les erreurs dues aux approximations des `float`, on peut vérifier si une valeur est **proche** d'une autre dans un intervalle autour d’elle.

````python
delta = 1e-5
var = 3.0 - 2.7

print(0.3 - delta < var < 0.3 + delta)
print(abs(var - 0.3) < delta)
````

````python
import math

var = 3.0 - 2.7
print(math.isclose(var, 0.3, abs_tol=1e-5))  # True
````
