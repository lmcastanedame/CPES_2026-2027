# Listes

## Définition

> Une **liste** est une structure de données qui contient une collection d’objets Python.  
> Il s’agit d’un nouveau type par rapport aux entiers, `float`, booléens et chaînes de caractères.
> On parle aussi d’**objet séquentiel** car une liste contient une séquence d’autres objets.

````python
animaux = ["girafe", "tigre", "singe", "souris"]
animaux
````

````python
mixte = ["girafe", 5, "souris", 0.15]
mixte
````

## Utilisation

Vous accédez à ses éléments par leur position. Ce numéro est appelé **indice** (ou *index*) de la liste.

````python
print(animaux[0])
print(animaux[1])
print(animaux[3])
````

````python
print(animaux[4])
````

## Opérations sur les listes

En Python, les listes supportent plusieurs types d'opérations comme :
- la **concaténation** (`+`) pour réunir plusieurs listes,
- la **duplication** (`*`) pour répéter le contenu d'une liste.

````python
ani1 = ["girafe", "tigre"]
ani2 = ["singe", "souris"]

print(ani1 + ani2)
print(ani1 * 3)
````

````python
liste1 = []
print(liste1)
liste1 = liste1 + [15]
print(liste1)
liste1 = liste1 + [-5]
print(liste1)
````

> On peut également ajouter des éléments **plus élégamment** avec la méthode `.append()`, qui est plus claire et efficace.

````python
liste1.append(13)
print(liste1)
liste1.append(-3)
print(liste1)
````

````python
print(ani2.append('chien'))
````

## Indisage négatif

En Python, il est possible d'accéder aux éléments d'une liste à l'aide d'**indices négatifs**, ce qui permet de compter à partir de la fin de la liste :

- `-1` : dernier élément
- `-2` : avant-dernier
- `-3` : etc.

````python
print(ani1)
ani1[-1]
````

## Tranches (slicing)

On peut extraire une portion (ou "tranche") d'une liste grâce à la syntaxe `liste[debut:fin]`, où `debut` est inclus et `fin` est exclu.  
Si on ne précise pas `debut` ou `fin`, Python considère respectivement le début ou la fin de la liste.

````python
animaux = ["girafe", "tigre", "singe", "souris"]

print(animaux[0:2])
````

````python
print(animaux[0:]) 
````

````python
print(animaux[:]) 
````

````python
print(animaux[1:])
````

````python
print(animaux[1:-1])
````

> On peut ajouter un troisième argument `pas` avec un second `:` pour sauter des éléments :

````python
print(animaux)
print(animaux[0:3:2])
````

## Fonctions

> La fonction `len()` permet de connaître **la longueur d'une liste**

````python
len(animaux)
````

> La fonction `range()` génère une séquence de nombres entiers dans un intervalle donné.  
Lorsqu'elle est combinée avec `list()`, on obtient une **liste** d'entiers.

````python
list(range(4))
````

````python
list('Bon       jour')
````

````python
list(range(10, 0, -1))
````

## Listes de listes

Il est tout à fait possible de créer des **listes contenant d'autres listes**.  
Cela permet de structurer les données de manière plus complexe.


````python
prairie1 = ["girafe", 4]
prairie2 = ["tigre", 2]
prairie3 = ["singe", 5]
savane = [prairie1, prairie2, prairie3]
savane
````

````python
savane[1] 
````

````python
print(savane[1][0])  
print(savane[1][1])  
````

## Minimum, maximum et somme d'une liste

````python
liste1 = list(range(10))
print(liste1)

print("Somme :", sum(liste1))
print("Minimum :", min(liste1))
print("Maximum :", max(liste1))
````

````python
liste2 = [2, 5, 6, 1]
sorted(liste2)
````

````python
sum(liste1)/len(liste1)
````

````python
sorted(animaux)
````

````python
min(['bonjour', 2, 5.6, 1+2j])
````

````python
min(['1', 3])
````

## Problème avec les copies de listes

Lorsque vous copiez une liste en Python avec `=`, ce n’est pas une **copie indépendante**, mais une **copie par référence**. Les deux variables pointent vers la **même** liste en mémoire.

````python
liste1 = list(range(10))
liste2 = liste1

liste1[2] = 89
liste2
````

````python
liste1 = list(range(10))

liste2 = liste1 *1

liste1[2] = 89
liste2
````

````python
liste1 = list(range(10))

liste2 = list(liste1)
liste1[2] = 89
liste2
````

````python
savane2 = list(savane)

savane[1][1] = 45
savane2[1][1]
````

````python
a = 1
list(a)
````

````python
import copy

str1 = 'Bonjour'
str2 = str1

liste1 = list(str1)
liste2 = copy.deepcopy(liste1)

liste1[1] = 'c'
print(liste2)
````
