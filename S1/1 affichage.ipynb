## Écriture formatée et `f-strings`

L'écriture formatée permet d'afficher des variables avec un format précis (ex. : justifié, avec un nombre de décimales, etc.).

Depuis Python 3.6, on peut utiliser les **f-strings** (`formatted string literals`), une méthode pour insérer des variables dans des chaînes de caractères.

```python
"Voici une chaîne classique"
```

```python
f"Voici une chaîne f-string"
```

```python
x = 32
nom = "John"

print(f"{nom} a {x} ans")
print(nom, "a", x, "ans")
```

```python
print(f"Bonjour {nom} ! Tu as {x} ans !")
```

```python
s = "Bonjour {nom}! Tu as {age} ans.".format(nom=nom, age=x)
s
```

```python
from math import pi

print(f"Pi vaut {pi:.2f}")
```

```python
x = 56
f"La valeur de {x} en binaire est {x:b}."
```

```python
x = 1234.56789

print(f"{x:.2f}")        # 2 décimales
print(f"{x:>10.2f}")     # Largeur 10, aligné à droite
print(f"{x:^10.2f}")     # Centré
print(f"{x:08.2f}")      # Remplissage par zéro
print(f"{255:x}")        # Hexa minuscule
print(f"{255:X}")        # Hexa MAJUSCULE
print(f"{0.75:%}")       # Pourcentage
```

```python
jour = 25
mois = 9
annee = 2024
f"{jour:0>2d}-{mois:0>2d}-{annee}"
```
