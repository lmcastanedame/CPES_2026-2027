# Variables

*Une **variable** est une zone de la mémoire de l'ordinateur dans laquelle une **valeur** est stockée. Aux yeux du programmeur, cette variable est définie par un **nom**, alors que pour l'ordinateur, il s'agit en fait d'une adresse, c'est-à-dire d'une zone particulière de la mémoire.*

## Déclaration et initialisation d’une variable en Python

En Python, la **déclaration** d'une variable et son **initialisation** (c’est-à-dire l’attribution de la première valeur) se font **en une seule étape**.

````python
x = 2
x
````

→ Typage dynamique

→ **Réservé un espace mémoire** pour stocker cette valeur

→ **Créé un lien** entre le nom `x` et la zone mémoire contenant `2`

> En C par exemple, il faudrait d'abord déclarer le type, puis initialiser la valeur.
>
> ```c
> int x;
> x = 2;
> ```

→ **Haut niveau** d’abstraction de Python

## Les types de variables

Une **variable** peut contenir différents **types de données**. 

- les **entiers** (`int`)
- les **nombres à virgule flottante** (`float`)
- les **chaînes de caractères** (`str` ou `string`)

On dit que Python est **fortement typé** mais avec un **typage dynamique** : le type est automatiquement détecté et associé à chaque valeur, sans qu'on doive le déclarer.

### Chaînes de caractères

- des guillemets simples `'...'`
- des guillemets doubles `"..."` 
- ou même avec trois guillemets `'''...'''` ou `"""..."""` (souvent utilisés pour des chaînes sur plusieurs lignes).

````python
a = "bonjour"
b = 'salut'
c = """girafe"""
d = '''lion'''

a, b, c, d
````

> ⚠️ **Séparateur décimal**

En Python, **le point `.` est utilisé pour les nombres à virgule**, pas la virgule `,` comme en français.

````python
x = 3.14
x
````

````python
y = 3,14
y
````

### Les variables booléennes

Un **booléen** est un type de variable qui ne peut prendre que **deux valeurs** : `True` (vrai) ou `False` (faux).

Ces deux mots (`True` et `False`) sont **réservés** en Python, c’est-à-dire qu’ils doivent toujours être écrits avec une majuscule au début.

````python
var = True
var2 = False

var, var2
````

> Les booléens sont très utiles pour les **conditions**, **tests logiques** et **structures de contrôle**

## Le nommage des variables

- Ils peuvent contenir des **lettres minuscules** (`a` à `z`), des **lettres majuscules** (`A` à `Z`), des **chiffres** (`0` à `9`) et le caractère `_`.
- Un nom **ne doit pas commencer** par un chiffre.
- Il est **déconseillé de commencer par un underscore** (cas avancés).
- **Pas d'espaces** dans les noms de variables.
- Évitez d'utiliser **des mots réservés** du langage Python (exemples : `print`, `range`, `for`, `from`, etc.).
- Python est **sensible à la casse** : `Nom`, `nom` et `NOM` sont trois variables différentes.
- Il est préférable d'utiliser des **noms explicites** pour améliorer la lisibilité du code.

````python
# Exemple valide
nom = "Alice"
Nom = "Bob"
_nom = "Charlie"
nom2 = "David"

print(nom, Nom, _nom, nom2)
````

````python
2nom = "Eve"
print(2nom)
````

````python
for = 5
````

## L’écriture scientifique

- `1e6` signifie : $1 \times 10^6$
- `3.12e-3` signifie : $3.12 \times 10^{-3}$


Remarques importantes :

- `1e6` ne signifie pas "le nombre exponentiel `e`" mais bien *puissance de 10*.
- Même si les parties gauche et droite du `e` sont des entiers, Python interprète toujours le résultat comme un `float`.

````python
print(1e6)
print(3.12e-3)
````

> Python permet d’utiliser le caractère `_` (underscore) pour **séparer les groupes de chiffres**

````python
avogadro_number = 6.022_140_76e23
print(avogadro_number)

humans_on_earth = 8_094_752_749
print(humans_on_earth)

print(80_94_7527.49)
````

## Les opérations

En Python, les **quatre opérations de base** (`+`, `-`, `*`, `/`) fonctionnent directement sur les types numériques (`int` et `float`).

````python
x = 45
x = x - 2
x
````

### Opérateurs combinés

- `-=` : soustraction puis affectation
- `*=` : multiplication puis affectation
- `/=` : division puis affectation

````python
x += 3 
x
````

````python
x /= 2
x
````

### Remarques :

- Si vous mélangez des `int` et des `float`, le résultat est **automatiquement** converti en `float`
- L’utilisation des **parenthèses** permet de gérer l’ordre d’évaluation des opérations (priorités)
- L’opérateur `/` renvoie **toujours** un `float`, même si le résultat semble être entier

> **Puissance :** L’opérateur `**` permet de calculer une puissance (exposant)

````python
2 ** 3
````

> **Division entière et modulo :** 

- `//` : division entière (quotient sans reste)
- `%` : modulo (reste de la division)

````python
5 // 4
````

````python
8 % 4
````

Les nombres de Friedman sont des nombres qui peuvent s'exprimer avec tous leurs chiffres dans une expression mathématique.

Par exemple, 347 est un nombre de Friedman, car il peut s'écrire sous la forme $4 + 7^3$. De même pour 127 qui peut s'écrire sous la forme $2^7 - 1$.

- $7 + 3^6$
- $(3 + 4)^3$
- $3^6 - 5$
- $(1 + 2^8) * 5$
- $(2 + 1^8)^7$
