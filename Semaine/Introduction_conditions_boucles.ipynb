# 1. Prendre des décisions avec Python

Un programme devient plus intéressant quand il peut choisir quoi faire selon une situation.

Pour cela, Python utilise des expressions booléennes : des expressions qui valent soit `True`, soit `False`.

```python
age = 20
print(age >= 18)
print(age < 18)
```

## Comparer des valeurs

Les opérateurs les plus courants sont :

- `==` : égal à
- `!=` : différent de
- `<` : inférieur à
- `<=` : inférieur ou égal à
- `>` : supérieur à
- `>=` : supérieur ou égal à

Attention : `=` sert à donner une valeur à une variable. Pour tester une égalité, on utilise `==`.

```python
mot_de_passe = "python"

print(mot_de_passe == "python")
print(mot_de_passe != "azerty")
```

## Combiner plusieurs tests

- `and` : les deux conditions doivent être vraies.
- `or` : au moins une condition doit être vraie.
- `not` : inverse une condition.

```python
age = 19
a_reservation = True

print(age >= 18 and a_reservation)
print(age < 18 or a_reservation)
print(not a_reservation)
```

## `if` : exécuter seulement si le test est vrai

Le bloc indenté sous `if` ne s'exécute que si la condition vaut `True`.

```python
temperature = 31

if temperature >= 30:
    print("Alerte chaleur")

print("Fin du contrôle")
```

## `if` / `else` : choisir entre deux chemins

```python
age = 15

if age >= 18:
    print("Tarif adulte")
else:
    print("Tarif réduit")
```

## `if` / `elif` / `else` : choisir parmi plusieurs cas

Python teste les conditions dans l'ordre. Dès qu'une condition est vraie, le bloc correspondant est exécuté.

```python
note = 14

if note >= 16:
    print("Très bien")
elif note >= 10:
    print("Validé")
elif note < 11:
    print("n")
else:
    print("À retravailler")
```

# 2. Répéter des actions avec `for`

Quand on connaît à l'avance les éléments à traiter, une boucle `for` est souvent le bon outil.

On peut parcourir une liste, une chaîne de caractères ou une suite de nombres créée avec `range()`.

```python
notes = [12, 15, 9, 18]

for note in notes:
    print("Note lue :", note)
```

## Utiliser `range()`

`range(5)` produit les nombres de 0 à 4.

`range(1, 6)` produit les nombres de 1 à 5.

```python
for numero in range(2, 5):
    print("Passage", numero)
```

## Accumuler un total

Une boucle peut construire progressivement un résultat.

```python
prix = [8, 12, 5, 20]
total = 0

for prix_article in prix:
    # total = total + prix_article
    total += prix_article

print("Total :", total, "euros")
```

# 3. Répéter jusqu'à ce qu'une situation arrive avec `while`

`for` est pratique quand on connaît la collection à parcourir.

`while` est pratique quand on veut répéter tant qu'une condition reste vraie, surtout quand on ne sait pas à l'avance combien de répétitions seront nécessaires.

```python
tentatives = 0

while tentatives < 3:
    print("Tentative", tentatives + 1)
    tentatives = tentatives + 1

print("Fin")
```

## Éviter les boucles infinies

Dans une boucle `while`, il faut vérifier que la condition pourra devenir fausse.

Sinon, le programme peut tourner sans fin.

## Activité : déboguer une boucle infinie

La cellule suivante est volontairement dangereuse : ne l'exécutez pas avant d'avoir trouvé le problème.

Que faut-il ajouter ou modifier ?

```python
compteur = 1

while compteur <= 5:
    print(compteur)
```

## Demander jusqu'à obtenir une réponse attendue

```python
choix = ""

while choix != "oui":
    choix = input("Tapez oui pour continuer : ")

print("Merci")
```

## `break` : sortir plus tôt d'une boucle

`break` peut servir à interrompre une boucle quand un événement arrive.

On l'utilise avec modération : souvent, une bonne condition dans `while` suffit.

```python
while True:
    reponse = input("Écrivez stop pour arrêter : ")
    if reponse == "stop":
        break

print("Boucle terminée")
```

# Récapitulatif

## Choisir entre `if`, `for` et `while`

- Utilisez `if` quand le programme doit choisir selon une condition.
- Utilisez `for` quand vous voulez parcourir une collection ou répéter un nombre connu de fois.
- Utilisez `while` quand vous voulez répéter jusqu'à ce qu'une condition change.

## Trois erreurs fréquentes à retenir

- Oublier les deux-points `:` après `if`, `elif`, `else`, `for` ou `while`.
- Confondre `=` et `==`.
- Oublier de modifier une variable dans une boucle `while`, ce qui peut créer une boucle infinie.
