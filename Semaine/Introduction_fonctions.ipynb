# 2. Donner des informations à une fonction : paramètres

Une fonction devient plus flexible quand on peut lui donner une information.

Cette information est reçue dans une variable placée entre les parenthèses.

```python
def saluer(prenom):
    print("Bonjour", prenom)

saluer("Lina")
saluer("Adam")
```

Dans `def saluer(prenom)`, `prenom` est le paramètre.

Dans `saluer("Lina")`, `"Lina"` est la valeur transmise à la fonction.

## Plusieurs paramètres

# 3. Produire un résultat : `return`

`print()` affiche une information à l'écran.

`return` renvoie une valeur au reste du programme.

Cette différence est essentielle : une valeur retournée peut être stockée, comparée, additionnée ou réutilisée.

```python
def additionner_et_afficher(a, b):
    print(a + b)

additionner_et_afficher(3, 4)
```

```python
def additionner(a, b):
    return a + b

resultat = additionner(3, 4)
print("Résultat :", resultat)
print("Résultat doublé :", resultat * 2)
```

# 4. Fonctions, conditions et boucles

Une fonction peut contenir une boucle.

Une boucle peut appeler une fonction.

Une fonction peut retourner une valeur utilisée plus tard par une condition.

```python
def est_reussie(note):
    return note >= 10

notes = [8, 12, 15, 6, 10]
compteur = 0

for note in notes:
    if est_reussie(note):
        compteur = compteur + 1

print("Notes validées :", compteur)
```

```python
def compter_reussites(notes):
    compteur = 0
    for note in notes:
        if note >= 10:
            compteur = compteur + 1
    return compteur

resultat = compter_reussites([8, 12, 15, 6, 10])
print("Nombre de réussites :", resultat)
```

# 5. Refactoriser un programme

Refactoriser signifie améliorer l'organisation du code sans changer ce que le programme fait.

On va partir d'un programme qui fonctionne, puis le découper progressivement.

## Programme de départ

Ce programme gère un petit carnet de notes. Il ne contient aucune fonction.

> Note enseignant : demander d'abord aux étudiants de lire le programme par blocs, sans chercher à tout modifier immédiatement.

```python
notes = []
choix = ""

while choix != "q":
    print("--- Carnet de notes ---")
    print("1 - Ajouter une note")
    print("2 - Afficher les notes")
    print("3 - Afficher la moyenne")
    print("q - Quitter")

    choix = input("Votre choix : ")

    if choix == "1":
        note = float(input("Note : "))
        if note >= 0 and note <= 20:
            notes.append(note)
            print("Note ajoutée")
        else:
            print("Note invalide")

    elif choix == "2":
        if len(notes) == 0:
            print("Aucune note")
        else:
            for note in notes:
                print(note)

    elif choix == "3":
        if len(notes) == 0:
            print("Aucune moyenne disponible")
        else:
            total = 0
            for note in notes:
                total = total + note
            moyenne = total / len(notes)
            print("Moyenne :", moyenne)

    elif choix == "q":
        print("Au revoir")

    else:
        print("Choix inconnu")
```

# Récapitulatif

## Questions essentielles

**À quoi sert une fonction ?**

À regrouper un bloc de code utile, à lui donner un nom, et à le réutiliser.

**Quand créer une fonction ?**

Quand un bloc se répète, quand une action a un rôle clair, ou quand le programme devient difficile à lire.

**Quelle est la différence entre un paramètre et un argument ?**

Le paramètre est le nom prévu dans la définition. L'argument est la valeur transmise lors de l'appel.

**Quelle est la différence entre `print()` et `return` ?**

`print()` affiche une valeur. `return` renvoie une valeur au programme.

**Quand une fonction doit-elle retourner une valeur ?**

Quand le reste du programme doit réutiliser le résultat.

## Quatre erreurs fréquentes

- Définir une fonction mais oublier de l'appeler.
- Oublier les deux-points ou l'indentation.
- Confondre `print()` et `return`.
- Oublier de transmettre un argument à une fonction qui attend un paramètre.
