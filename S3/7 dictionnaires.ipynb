# Les Dictionnaires

## Introduction

Les dictionnaires sont des structures de données qui permettent de stocker des couples **clé : valeur**.
Contrairement aux listes qui sont indexées par des entiers (0, 1, 2...), les dictionnaires sont indexés par des **clés** (souvent des chaînes de caractères).

```python
# Création d'un dictionnaire vide
mon_dico = {}

# Création avec des valeurs
contact = {
    "nom": "Dupont",
    "prenom": "Jean",
    "age": 30
}
print(contact)
```

## Accès et Modification

```python
# Accéder à une valeur
print(contact["prenom"])

# Modifier une valeur
contact["age"] = 31

# Ajouter une nouvelle paire clé:valeur
contact["ville"] = "Paris"

print(contact)
```

## Parcourir un dictionnaire
On peut parcourir les clés, les valeurs ou les deux.

```python
print("--- Clés ---")
for cle in contact.keys():
    print(cle)

print("\n--- Valeurs ---")
for valeur in contact.values():
    print(valeur)

print("\n--- Items (Clé, Valeur) ---")
for cle, valeur in contact.items():
    print(f"{cle} : {valeur}")
```
