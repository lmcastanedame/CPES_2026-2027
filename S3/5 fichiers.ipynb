````python
with open("animaux.txt", 'r') as filin:
    lignes = filin.readlines()
    for ligne in lignes:
        print(ligne)
filin.close()
````

## Méthode `.read()`

La méthode `.read()` lit **tout le contenu** du fichier d’un seul coup, et retourne **une seule chaîne de caractères**.

````python
with open("animaux.txt", "r") as filin:
    print(filin.read())
````

## Méthode `.readline()`

La méthode `.readline()` (sans `s`) permet de lire le fichier **ligne par ligne**. Chaque appel à `.readline()` lit la **ligne suivante** sous forme de chaîne.

> Elle est souvent utilisée avec une **boucle `while`**.

````python
with open("animaux.txt", "r") as filin:
    ligne = filin.readline()
    while ligne != "":
        print(ligne)
        ligne = filin.readline()
````

## Écriture dans un fichier

````python
animaux2 = ["poisson", "abeille", "chat"]

with open("animaux2.txt", "w") as filout:
    for animal in animaux2:
        filout.write(animal)
````

> Dans cet exemple, tous les mots sont écrits sur une seule ligne. Si l'on ouvre le fichier, on obtient ```poissonabeillechat```. Ce n'est pas toujours le résultat attendu. Pour écrire chaque animal sur une ligne différente, il faut ajouter un caractère de retour à la ligne (`\n`).

### Écriture avec retour à la ligne

````python
animaux2 = ["poisson", "abeille", "chat"]

with open("animaux2.txt", "w") as filout:
    for animal in animaux2:
        filout.write(f"{animal}\n")
````
