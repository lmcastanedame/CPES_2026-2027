# Fonctions

## Introduction

Jusqu'à présent, nous avons écrit des scripts où les instructions s'enchaînent les unes après les autres. Pour structurer nos programmes et éviter de répéter du code, nous utilisons des **fonctions**.

Une fonction est un bloc d'instructions nommé, qui peut prendre des paramètres et retourner un résultat.

### Définir une fonction simple

On utilise le mot-clé `def` suivi du nom de la fonction et de parenthèses.

```python
def dire_bonjour():
    print("Bonjour tout le monde !")
```

Pour exécuter la fonction, on l'appelle par son nom :

```python
dire_bonjour()
```

## Les arguments (paramètres)

On peut passer des informations à une fonction entre les parenthèses.

```python
def saluer(nom):
    print(f"Bonjour {nom} !")

saluer("Alice")
saluer("Bob")
```

On peut définir plusieurs arguments séparés par des virgules.

```python
def addition(a, b):
    print(f"La somme de {a} et {b} est {a + b}")

addition(5, 3)
```

## La valeur de retour (`return`)

Souvent, une fonction doit **renvoyer** un résultat pour qu'il soit utilisé ailleurs dans le programme, plutôt que de simplement l'afficher.
On utilise le mot-clé `return`.

```python
def carre(x):
    return x * x

resultat = carre(4)
print(f"Le carré de 4 est {resultat}")
```

> **Attention** : Dès que l'instruction `return` est exécutée, la fonction s'arrête.

## Portée des variables (Variables locales et globales)

Une variable définie à l'intérieur d'une fonction est **locale** : elle n'existe que dans cette fonction.
Une variable définie dans le programme principal est **globale**.

```python
def ma_fonction():
    x = 10
    y = 5
    print(f"Dans la fonction : x = {x}")

x = 5
ma_fonction()
print(f"En dehors : x = {x}")
```

## Récursivité

Une fonction peut s'appeler elle-même. C'est ce qu'on appelle la **récursivité**.
Il est essentiel d'avoir une **condition d'arrêt** pour éviter une boucle infinie.

```python
def compte_a_rebours(n):
    if n <= 0:
        print("Partez !")
    else:
        print(n)
        compte_a_rebours(n - 1)

compte_a_rebours(3)
```

### Exercice 1 : Parité
Écrire une fonction `est_pair(n)` qui prend un nombre entier `n` et renvoie `True` si le nombre est pair, et `False` sinon.

### Exercice 2 : Maximum
Écrire une fonction `maximum(a, b)` qui renvoie le plus grand des deux nombres `a` et `b` (sans utiliser la fonction `max()` de Python).

```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

# Test
print(f"Le max entre 10 et 20 est {maximum(10, 20)}")
```

### Exercice 4 : Factorielle (Récursif)
Écrire une fonction récursive `factorielle(n)` qui calcule $n! = n \times (n-1) \times ... \times 1$.
Rappel : $0! = 1$.

```python
def factorielle(n):
    if n == 0:
        return 1
    else:
        return n * factorielle(n - 1)

print(f"5! = {factorielle(5)}")
```

### Exercice 5 : Suite de Fibonacci (Récursif)
La suite de Fibonacci est définie par :
- $F_0 = 0$
- $F_1 = 1$
- $F_n = F_{n-1} + F_{n-2}$ pour $n \geq 2$.

Écrire une fonction récursive `fibonacci(n)` qui renvoie le $n$-ième terme de la suite.

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print("Les 10 premiers termes de la suite de Fibonacci :")
for i in range(10):
    print(fibonacci(i))
```

### Exercice 6 : Palindrome (Récursif)
Écrire une fonction récursive `est_palindrome(mot)` qui renvoie `True` si le mot est un palindrome (se lit de la même façon dans les deux sens), `False` sinon.
*Indice : Un mot est un palindrome si la première et la dernière lettre sont identiques et que le reste du mot (sans les extrémités) est aussi un palindrome.*

```python
def est_palindrome(mot):
    if len(mot) <= 1:
        return True
    if mot[0] != mot[-1]:
        return False
    return est_palindrome(mot[1:-1])

# Test
print(f"'radar' est un palindrome ? {est_palindrome('radar')}")
print(f"'python' est un palindrome ? {est_palindrome('python')}")
```

### Exercice 7 : Somme des chiffres (Récursif)
Écrire une fonction récursive `somme_chiffres(n)` qui prend un entier positif `n` et renvoie la somme de ses chiffres.
*Exemple : `somme_chiffres(123)` doit renvoyer `1 + 2 + 3 = 6`.*

```python
def somme_chiffres(n):
    if n < 10:
        return n
    else:
        return n % 10 + somme_chiffres(n // 10)

# Test
print(f"Somme des chiffres de 123 : {somme_chiffres(123)}")
print(f"Somme des chiffres de 9876 : {somme_chiffres(9876)}")
```

### Exercice 8 : Aplatir une liste (Récursif)
Écrire une fonction récursive `aplatir(liste)` qui prend une liste pouvant contenir des listes imbriquées et renvoie une seule liste plate contenant tous les éléments.
*Exemple : `aplatir([1, [2, 3], [4, [5]]])` doit renvoyer `[1, 2, 3, 4, 5]`.*

```python
def aplatir(liste):
    resultat = []
    for element in liste:
        if isinstance(element, list):
            resultat.extend(aplatir(element))
        else:
            resultat.append(element)
    return resultat

# Test
liste_imbriquee = [1, [2, 3], [4, [5]]]
print(f"Liste aplatie : {aplatir(liste_imbriquee)}")
```

### Exercice 9 : PGCD (Récursif)
Écrire une fonction récursive `pgcd(a, b)` qui renvoie le Plus Grand Commun Diviseur de `a` et `b` en utilisant l'algorithme d'Euclide.
*Rappel : `pgcd(a, b) = a` si `b = 0`, sinon `pgcd(b, a % b)`.*

```python
def pgcd(a, b):
    if b == 0:
        return a
    else:
        return pgcd(b, a % b)

# Test
print(f"PGCD de 48 et 18 : {pgcd(48, 18)}")
print(f"PGCD de 101 et 103 : {pgcd(101, 103)}")
```

### Exercice 10 : Les Tours de Hanoï (Récursif - Difficile)
Le but est de déplacer $n$ disques de la tour **A** vers la tour **C** en utilisant la tour **B** comme intermédiaire, en respectant les règles suivantes :
1. On ne peut déplacer qu'un seul disque à la fois.
2. On ne peut pas placer un disque plus grand sur un disque plus petit.

Écrire une fonction récursive `hanoi(n, source, cible, auxiliaire)` qui affiche les déplacements à effectuer (par exemple : "Déplacer le disque 1 de A vers C").

```python
def hanoi(n, source, cible, auxiliaire):
    if n == 1:
        print(f"Déplacer le disque 1 de {source} vers {cible}")
        return
    # Déplacer n-1 disques de source vers auxiliaire
    hanoi(n - 1, source, auxiliaire, cible)
    # Déplacer le disque n de source vers cible
    print(f"Déplacer le disque {n} de {source} vers {cible}")
    # Déplacer n-1 disques de auxiliaire vers cible
    hanoi(n - 1, auxiliaire, cible, source)

# Test pour 3 disques
print("Résolution pour 3 disques :")
hanoi(3, 'A', 'C', 'B')
```
